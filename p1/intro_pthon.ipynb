# Tipos de datos


* En Python **todo** es un objeto, por tanto también lo son todos los tipos de datos 
* Dos tipos generales de datos:
    * Escalares. Son objetos indivisibles (piensa en ellos como los átomos del lenguage)
        * int, float, complex, bool (True or False), None
    * Contenedores o también llamados _tipos estructurados_: tienen estructura interna, contienen un número arbitrario de enlaces a otros objetos
        * strings (str) Listas, tuplas, diccionarios, conjuntos

````python
a = 3             # objeto de tipo entero
g = True          # objeto de tipo booleano
l = [1, 2, 3]     # lista en la que sus tres elementos son enteros
b = [[1,2], "hola", 'adios']  #lista en la que su primer elemento es una lista y el segundo y tercero una cadena de caracteres
c = (1, 2, 3)     # tupla de enteros
d = { 'name' : "Fulano", 'last_name' : 'Perez', 'edad' : 18} # Diccionario, pares de llave-valor
````

* A diferencia de C las variables de Python _**no** están fuertemente tipadas_. Los tipos se asignan de forma implícita y se comprueban en tiempo de ejecución.

* Se puede verifica el tipo, clase, de un objeto con las funciones `type()` y `isinstance()` 

````python
print('tipo a:', type(a))      # Imprimimos el tipo del objeto a
print('tipo l:', type(l))      # Imprimemos el tipo del objeto l

print(isinstance(l, list))    # Preguntamos si el objeto l es una instancia de la clase list

if (type(l) is list):          # Condición lógica con el operador is
    print("Era de tipo lista")
````

# Tipos estructurados en Python

Python tiene cinco tipos estructurados: cadenas (`strings`), tuplas (`tuples`), listas (`lists`), diccionarios (`dicts`) y conjuntos (`sets`)

## String (cadenas):

* Son secuencias ordenadas de caracteres delimitadas por comillas simples, comillas dobles o triples


````python
a = 'esto es una cadena con el caracter 3'
b = """esto es otra cadena 
en dos lineas"""
c = "ESTO ES OTRA CADENA"
print(a)
print(b)
````

* Las cadenas de caracteres son **objetos indexables**. Por tanto sus elementos (caracteres) son accesibles mediante un índice. 

````python
print(a[0])    # accede al primer caracter de la cadena
print(a[-1])   # accede al ultimo caracter de la cadena
print(a[3:7])  # accede a los caracteres desde el cuarto al octavo
````

* Las cadenas son **objetos inmutables**. Una vez creado un objeto inmutable no se puede modificar. Además de las cadenas de caracteres, en python existen otros tipos primitivos que también son inmutables, por ejemplo, `int, float, tuple`también lo son.  El siguiente código lanza una excepción de la clase `TypeError` cuando se intenta modificar la cadena de caracteres

````python
a[0] = 3  # Error (se lanza una excepción) al intentar modificar el primer caracter de la cadena
````

* Las cadenas son objetos **iterables**, i.e., sus elementos se pueden iterar en un bucle

````python
a = "Hello World"

# Iterando sobre la cadena
# En cada paso de la iteración x toma el valor de un caracter de la cadena a 
for x in a:
    print(x)
````

````python
# Iterando sobre la cadena. Con el argumento modifico el caracter
# salto de carro que añade la función print 
for x in a:
    print(x, end = '')
````

### Operadores que _operan_ sobre cadenas de caracteres

|Operator |	Operation	| Description |
|-------- | ------------ | ----------- |
|Assignament|	s1 = s2 |	Link s2 to the object linked by s1. |
|Concatenation|	s1 + s2 |	Creates a new string concatenating two strings, s1 and s2. |
|Repetition |	s * n	| Makes n copies of string, s.|
|Indexing	| s [ i ] |	Indexing a string, returning element at index i. |
|Slicing |	s[ i : j : step] 	| Slicing a string from i to j with an optional step.|
|Membership |	x in s |	Returns True if the element, x is in the string, s. Otherwise False.|
||x not in s | 	Returns True if the element, x is not in the string, x. Otherwise False.|
|Comparison |	s1 == s2 | 	Returns True if string, s1 is the same as string, s2. Otherwise False.|
 || s1 != s2 | 	Returns True if string, s1 is not the same as string, s2. Otherwise False.|

## Listas

### Introducción


Secuencia mutable, ordenada e iterable de **enlaces a objetos** (los elementos) posiblemente de diferentes tipos. Por tanto:
* sus elementos son accesible mediante un índice.
* se pueden obtener *trozos* de una lista (**Cuidado!!** creas una nueva lista donde las posiciones enlazan a los mismos objetos de la lista original)
* se puede iterar para obtener sus elementos

````python
l = ["Hello world", 1, 2, [3,4], 4, 5, 6]
print(l)      # imprimo la lista
print(l[0])   # imprimo el objeto enlazado por primer elemento de la lista
print(l[-1])  # imprimo el objeto enlazado por el ultimo elemento de la lista
print()

l2 = l[:4]  # Crea un nuevo objeto de tipo lista con cuatro enlaces. 
            #   Las posiciones de la nueva lista enlazan a los cuatro primeros objetos 
            #   enlazados por la lista l

print(l2)    
print(l[3][0])
````

 * Las listas son objetos iterables (el objeto lista implementa el método mágico `__iter__`) y por tanto pueden iterarse en un bucle

````python
for item in l:
    print(item)
````

* Las Listas se pueden concatenarse con el operador  `+`

````python
l = ["Hello world", 1, 2, [3,4]]
l3 = ["Bye"] 
l = l[:2] +  l3
l
````

* `l =[ ]` crea una lista vacía. Es equivalente a `l = list()`

* `len(obj)` devuelve el número de elementos en la lista `obj` . La función *len* cuando se invoca sobre un objeto *ejecuta* el método mágico `__len__` de dicho objeto. 
* El operador `*` aplicado a una lista concatena la lista consigo misma tantas veces como se indique en el coeficiente   

````python
l = 3 * [1,2]     # crea una lista concatenando 3 veces la lista [1,2]
print(l)
print(len(l))
````

* La función `list(a)` crea una lista a partir del objeto iterable a

````python
a = "cadena"  # cadena
b = list(a)
print(b)
````

````python
d = {'nombre':'Eduardo', 'edad':27}  # diccionario
b = list(d)                          # crea una lista con las llaves del diccionario
print(b)
````

* Cuando `list()`no reciba ningún argumento creará una lista vacía 

````python
l = list()
print(l)
````

* Las listas se pueden *desempaquetar*

````python
x, y, z = ["Hello", 'World', [1,2,3]]
print(x, y, z)
print(type(x), type(y), type(z))
````

    * Añadir o eliminar elementos del final de la lista es rápido $O^*(1)$, i.e., $O(1)$ en coste amortizado.
    * Añadir o eliminar elementos de otras posiciones de la lista no es eficiente, $O(n)$. Por tanto la lista es una estructura de datos eficiente para implementar pilas pero no para colas.

### Algunos métodos de las listas

|Método          |Decripción                               |Peor caso, $^*$ coste amortizado       |
|:--------------|:------------------------------------------ | -----:|
|L.append(e)   |añade el objeto e al final de la lista L | $O^*(1)$|
|L.pop()   |elimina y devuelve el último elemento de la lista L. Lanza una excepción si L está vacía | $O^*(1)$|
|L.pop(i)   |elimina y devuelve el elemento que ocupa la posición i de la lista L. Lanza una excepción si L está vacía | $O(n)$|
|L.count(e)    |retorna le número de veces que aprece el objeto `e` en la lista L | $O(n)$|
|L.insert(i, e)|añade el objeto e en la lista L en la posición i | $O(n)$|
|L.remove(e)|elimina la primera ocurrencia del objeto e en la lista L| $O(n)$|
|L.extend(L1)|añade los elementos de la lista L1 al final de la lista L | $O(k)$|

> * Hacer una copia de la lista `L1` antes de entrar en el bucle y cada vez que tuviese que eliminar un elemento lo eliminase de la lista-copia y no de la original. Pero ¿Cómo hacer una copia de una lista?

> Debes distinguir entre lo que es una referencia a un objeto, una copia superficial (**shallow copy**) y una copia profunda (**deep copy**). Analiza el siguiente [pythontutor](https://pythontutor.com/render.html#code=%0AL1%20%3D%20%5B1,%202,%203,%204%5D%0AL_ref%20%3D%20L1%20%20%23%20esto%20no%20es%20una%20copia%20ya%20que%20L_ref%20enlaza%20con%20el%20mismo%20objeto%20que%20L1%0A%0AL_shallow%20%3D%20L1%5B%3A%5D%20%23%20copia%20shallow%20se%20copian%20los%20elementos%20pero%20no%20los%20enlaces%20de%20estos%0AL1%20%3D%20%5B%5B1,%202%5D,%203,%204%5D%0AL_shallow%20%3D%20L1%5B%3A%5D%0A%0A%0Aimport%20copy%0AL_deep%20%3D%20copy.deepcopy%28L1%29&cumulative=false&curInstr=7&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=3&rawInputLstJSON=%5B%5D&textReferences=false)

````python
import copy   # importo el modulo de la librería estandard

def remove_dup(L1, L2):
    '''Assumes that L1 and L2 are lists.
    Removes any element from L1 that also occurs in L2
    No muy buena solución'''
    
    L_copy = copy.deepcopy(L1)
    for ele in L1:
        if ele in L2:
            L_copy.remove(ele)

    return L_copy 

# main program 
L = [1, 2, 3, 4]      
Ld = [1, 2, 5, 6]

L_filter = remove_dup(L, Ld)
print(L_filter)
````

> * Una solución mucho mejor (¿sabrías decir porqué?) sería crear una nueva lista vacía y si el elemento de L1 no estuviese en L2 añadirlo a la nueva lista 

````python
def remove_dup(L1, L2):
    '''Assumes that L1 and L2 are lists.
    Removes any element from L1 that also occurs in L2
    Mejor solución'''
    
    L_ = []                      # Creo una lista vacía
    for ele in L1:               #   recorro uno a uno la lista L1
        if ele not in L2:
            L_.append(ele)

    return L_ 

# main program 
L = [1, 2, 3, 4]      
Ld = [1, 2, 5, 6]

L_filter = remove_dup(L, Ld)
print(L_filter)
````

### Introducción a iteradores: funciones _zip()_,  _enumerate()_ y _range()_ 
 #### Función _zip()_

La función `zip()` devuelve un objeto de tipo `zip`. Un objeto zip es un `iterador` de tuplas. Un iterador es un objeto que permite generar objetos uno a uno para, eventualmente, operar con ellos. Para que un objetos sea un iterador debe implementar el método `next()`.

* Los parámetros que se le pasen a la función `zip()` deben ser obejetos iterables:
    * Sino no se le pasasa ningún parámetro, `zip()` devuelve un iterador vacío
    * Si se le pasa un único prámetro, `zip()` devuelve un iterador de tuplas, donde cada tupla tendrá un único elemento.
    * Si le pasas como argumento varios objetos iterables (por ehjemplo dos listas o dos objetos de tipo range) `zip()` devuelve un iterador de tuplas donde cada tupla tiene un elemento proveniente de cada uno de los iterables pasados como argumentos.

````python
l1 = ['C', 'Python', 'JavaScript'] 
l2 = [17, 30, 40]
l3 = ['first', 'second', 'third', 'four']

# Creamos el iterador con la función zip pasándole como argumento tres listas
z = zip(l1, l2, l3) 
print(z)
````

* Una vez creado el iterador, la función next nos permite iterar el iterador obteniendo uno a uno sus objetos.

````python
print(next(z))
print(next(z))
print(next(z))
````

El objeto devueltos en cada iteración del iterador `z` es una tupla con los correspondientes elementos de los argumentos con las que se creo `z`, en este caso tres listas. La primera vez que iteramos obtenemos  una tupla formada por los tres _primeros_ elementos de las tres listas. La segunda vez con los tres _segundos_ y la tercera vez con tres _terceros_. 
Si volviésemos a iterar `z` obtendríamos una excepción de la clase `StopIteration` porqué el iterador no puede generar mas objetos. 

````python
next(z)
````

* Nota como el iterador genera tantos objetos como elementos tenga el menor de los iterables que se han utilizado para crear el objeto zip. En este caso la longitud del menor de los iterables utilizados era tres. 

* Aunque en el ejemplo anterior utilizamos la función `next()` para iterar, habitualmente un iteardor se itera con un bucle `for`

````python
l1 = ['C', 'Python', 'JavaScript'] 
l2 = (1972, 2000, 1995)
l3 = ['Bell Labs', 'Centrum Wiskunde & Informatica', 'Netscape', 'Microsoft']

for item in zip(l1, l2, l3):
    print(item)
````

#### ¿iterador o un Iterable?
 
* ¿Cómo podríamos saber si un objeto es un iterador o es iterable? Una posibilidad sería importar el módulo collections y preguntar si el objeto es una instancia de la clase `collections.abc.Iterator` o de la clase `collections.abc.Iterable`
* ¿Se te ocurre otra posibilidad?

````python
import collections.abc as co

l = ['hola', 'don', 'pepito']             # Iterable pero no iterador
ra = range(len(l))                        # Iterable pero no iterador (on fly == lazy)
z = zip(l, r)                             # Iterador
e = enumerate(l)                          # Iterador
ge = (n**2 for n in range(3))             # Iterador (expression generator)
lc = [n**2 for n in range(3)]             # Lista por tanto iterable 
rev = reversed(l)

print('List:', isinstance(l, co.Iterable), isinstance(l, co.Iterator) ) 
print('range:', isinstance(ra, co.Iterable), isinstance(ra, co.Iterator) ) 
print('zip:', isinstance(z, co.Iterable), isinstance(z, co.Iterator) ) 
print('enumerate:', isinstance(e, co.Iterable), isinstance(e, co.Iterator) ) 
print('expression_generatos:', isinstance(ge, co.Iterable), isinstance(ge, co.Iterator) ) 
print('list_comprenhesion:', isinstance(lc, co.Iterable), isinstance(lc, co.Iterator) ) 
print('reversed_list:', isinstance(rev, co.Iterable), isinstance(rev, co.Iterator) ) 

# Puedes iterar un iterador con la funcion next()
next(rev)
````

>

>  * Solución **sin** dictionary comprenhesion. Podemos crear un nuevo diccionario copia del anterior y modificar sus valores con un bucle `for`

````python
price_dollar = {'milk': 1.02, 'coffee': 2.5, 'bread': 2.5}      # item price in dollars
conversion = 0.99  # 1$ equivalent to 0.99€

price_euro = dict(price_dollar)                                 # new (soft) copy
conversion = 0.99                                               # 1$ equivalent to 0.99€            

for item, value in price_euro.items():
    price_euro[item] = value * conversion

print(price_euro)    
````

> * Solución mas compacta **utilizando _dictionary comprenhesion_**

````python
price_euro = {item: value * conversion for item, value in price_dollar.items()}
print(type(price_euro))
print (price_euro)
````

Como `l` enlaza a un objeto mutable, este puede ser modificado _dentro_ de la función y su _efecto_ se observa fuera del ámbito de la función.

Supongamos que tenemos una función (o funciones como en este caso) que tiene un único parámetro de entrada. Por ejemplo, las siguientes funciones calculan la *potencia al cuadrado* y la *potencia a la tres* del valor que se pasa a la función.

````python
def cubic (x):
    return x**3

def square (x):
    return x**2

#-------------
print(cubic(2))
print(square(2))
````

Queremos crear una función `apply(l, f)`con dos parámetros de entrada, uno de tipo lista, l,  y otro que sea una función,f. La función `apply()` debe retornar una lista cuyos elementos sean el resultado de aplicar f a cada uno de los elementos de la lista `l`.

````python
def apply(l, f):
    return [f(i) for i in l]  # list comprenhesion

# Driver program
l =  [4, 9, 25, 36]
print (apply(l, square) )
print (apply(l, cubic))
````

Nuestra función `apply()` es correcta, pero  ¿y si quisiésemos invocar a `apply()` con una función que requiere parámetros adicionales? Por ejemplo queremos *aplicar* la función `power()` o la función `oper()` a los elementos de una lista ¿Qúe puedo hacer?. 

````python
def power(base, exp_):
    return base**exp_

def oper(coef, base, exp_):
    return coef*(base**exp_)
````

Tengo, al menos, las siguientes alternativas:

 * Invocar `apply()` pasándole como argumento una función en notación lambda (ver el apartado de funciones `lambda`) que, a su vez, invoca a la función que quiero *aplicar* a los elementos de la lista. A la función lambada le paso los paraémetros adecuados.

````python
l =  [4, 9, 25, 100]
print(apply(l, lambda x: power(x, 1./2.)))
print(apply(l, lambda x, y = 2, z = 3: oper(x, y, z)))
print(apply(l, lambda x, y = 2, z = 2: oper(x, y, z)))
````

   * Definir la función `apply()` como una función con _un número variable de parametros_. La función incluye un parámetro *tipo* `*arg` (ver apartado a continuación).
   
   *Nota: Informalmente diremos que el parámetro _arg_ recibira una lista desempaquetada*

````python
def apply(lista, f, *args):
    return [f(x, *args) for x in lista]

# Driver program
l =  [4, 9, 25, 36]
print(apply(l, power, 1/2))
print(apply(l, square))
print(apply(l, oper, 2, 2))
````

## Argumentos *args* y *kwargs*

### En la invocación de una función

* En el siguiente ejemplo la función `fun()` tiene tres argumentos.  Podemos invocarla pasándole la lista `args`en la que, previamente, hemos desempaquetado sus elementos con `*`  

````python
def fun(x, y, oper):
    ''' Funcion de tres argumentos'''
    if oper == '+':
        return x+y
    else:
        return x*y
        
args = (1, 2, '*')

a = fun(1, 2, '*')
print(a)

args = (1, 2, '*')
b = fun(args[0], args[1], args[2])
print(b)

c = fun (*args)  # equivalente a ejecutar: fun(args[0], args[1], args[2])
print(c)
````

* La siguiente función tiene dos argumentos posicionales (`x`e `y`) y uno por defecto (`tipo`). Podemos invocara a la función con una tupla que desempaquetamos para asignar valor a los argumentos posicionales y un diccionario que, a su vez, desempaquetamos para asignar los argumentos con clave. 

````python
def fun (x, y, tipo = 'triangulo'):
    if tipo == 'triangulo':
        return x*y/2
    else:
        return x*y

args = (4,2)
kwargs = { 'tipo' : "rectangulo"}
fun(*args, **kwargs) # equivalente a ejecutar: fun (args[0], args[1], tipo="rectangulo")
````

* La función anterior también podemos invocarla usando únicamente un diccionario que desempaquetamos

````python
kwargs = {'y':2, 'x':4,  'tipo' : "rectangulo"}
fun(**kwargs) # equivalente a ejecutar: fun (args[0], args[1], "rectangulo")
````

o una única tupla

````python
kwargs = (4, 2, "rectangulo")
fun (*args)
````

### En la definición de la función

* Podemos definir una función con un número arbitrario de parámetros posicionales mediante el uso de `args`. En el ejemplo, la variable args se enlaza a una tupla tal y como puedes observar en  [pythontutor](https://pythontutor.com/render.html#code=def%20add_arg_var%20%28*args%29%3A%0A%20%20%20%20value%20%3D%200%0A%20%20%20%20for%20i%20in%20args%3A%0A%20%20%20%20%20%20%20%20value%20%2B%3D%20i%0A%20%20%20%20print%20%28value%29%0A%20%20%20%20%0A%0Aadd_arg_var%20%283,4,5%29%0Aadd_arg_var%20%283,4,5,6%29&cumulative=false&curInstr=15&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=3&rawInputLstJSON=%5B%5D&textReferences=false)

````python
import math

def potencia(x, exponente = 2):
    return x**exponente
    
def logaritmo(x, base = 10):
    return math.log(x, base)

def square(x):
    return x**2

def operacion(x, f, **kwargs):  # en la definición de la función
    ''' Version 1'''
    return f(x, **kwargs)        # en la invocacion tiene que pasarse los argumentos desempaquetados

# Drive code
print(operacion (x=10, f=potencia, exponente = 3) )
print(operacion (x=10, f=logaritmo, base = 10) )
print(operacion (x=10, f=square) )
````

* La función `operacion()` podríamos programarla de otra forma. En este caso `operacion()` tiene un argumento con clave `param` cuyo valor por defecto es una lista vacía. Cuando `operacion()` invoca a `f` desempaqueta la tupla `param`   

````python
def operacion(x, f, param = tuple()):
    ''' Version 2 de la función operacion '''
    return f(x, *param)


# Drive code
print(operacion(x=10, f=potencia, param=(3,) ) )
print(operacion(x=10, f=logaritmo, param=(2,) ) )
print(operacion(x=10, f=square) )
````
